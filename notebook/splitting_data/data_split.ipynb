# Data Splitting (Stratified Holdout Strategy with Validation Set)

## Stratification is based on the target variable

### **Import Dependecies**

```python
import pandas as pd
from pathlib import Path
# from sklearn.model_selection import train_test_split
from src.utils.data_split.functions import (
    split_data,
    create_y_class_report,
    export_split_data_to_csv,
    binarize_readmitted
)
```

### **Import the dataset:**

```python
# initializing the paths
PATH = Path("../../data/source/diabetic_data.csv")
# used for exporting the splitted datasets
SPLITTED_PATH = Path("../../data/interim/split_stratified")
 #reading the datasets
df = pd.read_csv(PATH)
```

**Binarize Target Variable (readmission to hospital):**

- 'No readmission' is mapped to 0,
- '>30' days is mapped to 1,
- and '<30' days is mapped to 1.

```python
df = binarize_readmitted(df)
df[['encounter_id', 'readmitted']].head()
```

### **Applying the  splitting functions**

**Applying the data split function**

```python
X_train_for_cv, X_train_mini, X_val, X_test, \
    y_train_for_cv, y_train_mini, y_val, y_test = split_data(df,
                                                      test_size_step_1= 0.2,
                                                      test_size_step_2= 0.25)
```

**Brief report of target class counts and percentages for each dataset**:

```python
y_class_report = create_y_class_report(y_train_mini, y_val, y_test)
y_class_report
```

**Export datasets to ../data/interim/splitted/**

```python
split_file_paths = export_split_data_to_csv(
    X_train_for_cv, X_train_mini, X_val, X_test,
    y_train_for_cv, y_train_mini, y_val, y_test,
    output_folder= SPLITTED_PATH
)
```
