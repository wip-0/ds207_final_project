# Data Splitting (Group Shuffle Split by group (Patient number))

#### **Import Dependecies**

```python
import pandas as pd
from pathlib import Path
# from sklearn.model_selection import train_test_split, GroupShuffleSplit
from src.utils.data_split.functions import (split_data_by_patient,
                                            create_y_class_report,
                                            export_split_data_to_csv,
                                            binarize_readmitted
                                            )
```

#### **Import the dataset:**

```python
# initializing the paths
PATH = Path("../../data/source/diabetic_data.csv")
# used for exporting the splitted datasets
SPLITTED_PATH = Path("../../data/interim/split_groupshuffle")
 #reading the datasets
df = pd.read_csv(PATH)
df.head()
```

- '>30' days is mapped to 1,
